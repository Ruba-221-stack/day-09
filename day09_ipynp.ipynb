## Task 1

```python
import numpy as np

# البيانات الأصلية
salaries = np.array([6000, 6500, 7000, 7200, 7500, 8000, 95000])

print("=== Task 1 ===")

# المتوسط والوسيط
mean_salary = np.mean(salaries)
median_salary = np.median(salaries)

print("\n1- Mean:", mean_salary)
print("2- Median:", median_salary)

# الربع الأول والثالث
Q1 = np.percentile(salaries, 25)
Q3 = np.percentile(salaries, 75)

print("\n3- Q1:", Q1)
print("4- Q3:", Q3)

# IQR
IQR = Q3 - Q1
print("5- IQR:", IQR)

# حدود القيم الشاذة
lower_bound = Q1 - (1.5 * IQR)
upper_bound = Q3 + (1.5 * IQR)

print("\n6- Lower Bound:", lower_bound)
print("7- Upper Bound:", upper_bound)

# القيم الشاذة
outliers = salaries[(salaries < lower_bound) | (salaries > upper_bound)]
print("8- Outliers:", outliers)

# حذف المدير وإعادة الحساب
salaries_without_manager = salaries[salaries != 95000]

new_mean = np.mean(salaries_without_manager)
new_median = np.median(salaries_without_manager)

print("\n=== After Removing Manager Salary ===")
print("New Mean:", new_mean)
print("New Median:", new_median)

# حساب Z-Score
mean = np.mean(salaries)
std = np.std(salaries)

z_scores = (salaries - mean) / std

print("\n=== Z-Scores ===")
for salary, z in zip(salaries, z_scores):
    print(f"Salary = {salary}, Z-Score = {z:.2f}")
```

## Task 3

```python
import numpy as np

# قراءة البيانات
data = np.genfromtxt(
    "insurance.csv",
    delimiter=",",
    skip_header=1,
    usecols=(6)
)

charges = data

# حساب الربع الأول والثالث
Q1 = np.percentile(charges, 25)
Q3 = np.percentile(charges, 75)

# حساب IQR
IQR = Q3 - Q1

# الحدود
lower_bound = Q1 - (1.5 * IQR)
upper_bound = Q3 + (1.5 * IQR)

# استخراج القيم الشاذة
outliers = charges[(charges < lower_bound) | (charges > upper_bound)]

print("Q1 =", Q1)
print("Q3 =", Q3)
print("IQR =", IQR)
print("Lower Bound =", lower_bound)
print("Upper Bound =", upper_bound)
print("Number of Outliers =", len(outliers))
print("First 10 Outliers:")
print(outliers[:10])

# نسبة القيم الشاذة
outlier_percentage = (len(outliers) / len(charges)) * 100
print(f"\nOutlier Percentage: {outlier_percentage:.2f}%")
```

## Task 4

```python
import numpy as np

# قراءة الأعمدة المطلوبة
data = np.genfromtxt(
    "insurance.csv",
    delimiter=",",
    skip_header=1,
    usecols=(0, 2, 3, 6)
)

age = data[:, 0]
bmi = data[:, 1]
children = data[:, 2]
charges = data[:, 3]

# دالة Min-Max Normalization
def normalize(x):
    return (x - np.min(x)) / (np.max(x) - np.min(x))

# تطبيق التطبيع
age_norm = normalize(age)
bmi_norm = normalize(bmi)
children_norm = normalize(children)
charges_norm = normalize(charges)

# عرض النتائج
print("First 5 normalized values:\n")

print("Age:")
print(age_norm[:5])

print("\nBMI:")
print(bmi_norm[:5])

print("\nChildren:")
print(children_norm[:5])

print("\nCharges:")
print(charges_norm[:5])

print("\nRanges After Normalization:")
print("Age:", np.min(age_norm), "to", np.max(age_norm))
print("BMI:", np.min(bmi_norm), "to", np.max(bmi_norm))
print("Children:", np.min(children_norm), "to", np.max(children_norm))
print("Charges:", np.min(charges_norm), "to", np.max(charges_norm))
```

## Task 5

```python
import numpy as np

# إنشاء البيانات
rng = np.random.default_rng(7)

brightness = 1000 + rng.normal(0, 1, 240)

# عمليات العبور المخفية
for start in [30, 102, 174]:
    brightness[start:start + 5] -= 20

# عدد القراءات
print("عدد القراءات:", len(brightness))

# المتوسط والوسيط
mean_brightness = np.mean(brightness)
median_brightness = np.median(brightness)

print("Mean =", mean_brightness)
print("Median =", median_brightness)

# حساب Z-Score
z_scores = (brightness - mean_brightness) / np.std(brightness)

# القراءات المنخفضة
mask = z_scores < -3

print("\nعدد القراءات الشاذة:", np.sum(mask))
print("الساعات الشاذة:", np.where(mask)[0])

# تحديد بداية كل عبور
transit_starts = np.where(mask & np.concatenate(([True], ~mask[:-1])))[0]

print("\nبداية العبور:", transit_starts)

# الزمن بين كل عبورين
period_hours = np.diff(transit_starts)
period_days = period_hours / 24

print("\nالفترة بين العبورات (بالساعات):", period_hours)
print("الفترة بين العبورات (بالأيام):", period_days)

# حساب مقدار الانخفاض
normal_level = np.median(brightness[~mask])
transit_level = np.mean(brightness[mask])

drop_amount = normal_level - transit_level
drop_percentage = (drop_amount / normal_level) * 100

print("\nمقدار الانخفاض =", drop_amount)
print("نسبة الانخفاض (%) =", drop_percentage)
```
